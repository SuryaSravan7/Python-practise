PALINDROME

```python
n= int(input("Enter the Number: "))
i=n
rev=0
while i>0:
  rem=i%10
  rev=rev*10+rem
  i=i//10
if rev==n:
  print("Yes it is a palindrome")
else:
  print("Nope it is not a palindrome")
```
