```python
year= int(input("Enter the year: "))
if year%4==0 and year%100!=0:
  print("leap year")
elif year%100==0 and year%400==0:
  print("century leap year")
else:
  print("not a leap year")
```

```python
print("1 for add,2 for sub,3 for division,4 for float,5 for mul,6 for power,7 for modulus")
n1=int(input("Enter the Number: "))
n2=int(input("Enter the Number: "))
operation= int(input("Enter the operation: "))
if operation==1:
  print(n1+n2)
elif operation==2:
  print("the sub is",n1-n2)
elif operation==3:
  print("the div is",n1/n2)
elif operation==4:
  print("the float is",n1//n2)
elif operation==5:
  print("the mul is",n1*n2)
elif operation==6:
  print("the power is",n1**n2)
elif operation==7:
  print("the modulus is",n1%n2)
else:
  print("Wrong choice")
```
